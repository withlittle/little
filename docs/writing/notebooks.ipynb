# Notebooks

A Jupyter notebook (`.ipynb`) in `docs/` is a page, as it was saved, outputs
and all, without Jupyter or Python, or converting it first. This page is
one, `docs/writing/notebooks.ipynb`: its Markdown cells are Markdown, its
code cells are code blocks, highlighted in the notebook's language, and each
one's outputs follow it.

## Code, and what it printed

What the code printed, and a result that's only text, are code blocks in
`output`, which aren't highlighted. Terminal colours are left out, and a
progress bar shows as it ended.

```python
words = ["little", "builds", "notebooks"]
for word in words:
    print(word.upper())
```

```python
sum(len(word) for word in words)
```

## Tables

A table pandas shows is its HTML, so it needs [HTML](markdown.md#html) in
Markdown, which is on unless you turn it off. Without it, a table shows its
text.

```python
import pandas as pd

files = pd.DataFrame({
    "kind": ["Markdown", "Notebook", "Image"],
    "files": [5, 1, 1],
})
files
```

## Charts and images

A chart is an image in the page itself, as a `data:` URL, so it needs no file
beside the notebook. So is an image attached to a Markdown cell, as Jupyter
shows it, with ``, though not with an HTML
`<img>`. A chart in SVG, like this one, needs HTML too: without it, a chart
shows its PNG, if it has one, or its text.

```python
files.plot.barh(x="kind", y="files", legend=False);
```

## Writing your own

Save a notebook with its outputs in `docs/`, and it's a page. Its title,
table of contents, links, math, like $e^{i\pi} + 1 = 0$, diagrams, and
footnotes work as in a Markdown file, but:

- its frontmatter goes in a raw or Markdown cell at its top, as Quarto and
  MyST put it, and as this one's is. Other raw cells are left out;
- a cell's footnotes are its own: write each note in the cell that refers
  to it, and give notes in different cells different names;
- an output that needs Jupyter's JavaScript to show it, like a widget, shows
  its text, or nothing;
- code blocks in its Markdown cells, and in Markdown a cell showed, go to
  [plugins](../plugins/using-plugins.md), but its code cells don't.

A link to a notebook, like `[Notebooks](notebooks.ipynb)`, leads to its
page, and a folder's `README.ipynb` is its landing page, as a `README.md`
is. little reads version 4 of Jupyter's format, which Jupyter has saved
since 2015: open an older notebook in Jupyter, and save it, to upgrade it.

[Styling Markdown](../themes/styling-markdown.md) shows the HTML each
output is, for a theme's CSS.
