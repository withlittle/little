## Tables

A table pandas shows is its HTML, as long as the site's Markdown can
have HTML in it, as it can unless `raw_html` is off.

```python
import pandas as pd

files = pd.DataFrame({
    "kind": ["Markdown", "Notebook", "Image"],
    "files": [5, 1, 1],
})
files
```

## Charts

A chart is an image in the page itself, so it needs no file beside the
notebook.

```python
files.plot.barh(x="kind", y="files", legend=False);
```
